```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
```

```python
def normalization(X):
    return (X/127.5) - 1

datagen = ImageDataGenerator(preprocessing_function=normalization)
train_it = datagen.flow_from_directory('dataset/test_set/', class_mode='binary', batch_size=64, target_size=(64,64),shuffle=False)
print(train_it.class_indices)
print(train_it.labels)
print(train_it.next()[0][0])
```

```python
def load_i():
    img = load_img('dataset/test_set/cats/cat.4001.jpg', target_size=(64,64))
    img = img_to_array(img)
    # print(img)
    img = normalization(img)
    img = img.reshape(1,64,64,3)
    return img

img = load_i()
print(img) 
```
